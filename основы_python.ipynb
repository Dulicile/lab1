#Задача 1
Написать функцию на вход которой подается строка, состоящая из латинских букв.
Функция должна вернуть количество гласных букв (a, e, i, o, u) в этой строке. Написать тесты для кода

```python
def count_vowel(line):
    vowel = 'aeiou'
    count = 0 
    line_lower = line.lower()
    for char in line_lower:
        if char in vowel:
            count += 1
    return count 
    
test1 = 'Hello, my name is Yaroslava'
test2 = 'aeiou'
test3 = 'bcdgh'
test4 = ' '
test5 = 'AEIOU'
print(count_vowel(test1) == 9)
print(count_vowel(test2) == 5)
print(count_vowel(test3) == 0)
print(count_vowel(test4) == 0)
print(count_vowel(test5) == 5)
```

# Задача 2
Написать функцию на вход, которой подается строка. Функция должна вернуть true, если
каждый символ в строке встречается только 1 раз, иначе должна вернуть false. Написать тесты для кода

```python
def not_repeat(line):
    chars = set()
    for char in line:
        if char in chars:
            return False
        chars.add(char)
    return True

test1 = 'abcde'
test2 = 'aaaaaa'
test3 = 'aA'
test4 = 'hello'
print(not_repeat(test1) == True)
print(not_repeat(test2) == False)
print(not_repeat(test3) == True)
print(not_repeat(test4) == False)
```

# Задача 3
Написать функцию, которая принимает положительное число и возвращает количество
бит равных 1 в этом числе. Написать тесты для кода

```python
def count_bit(number):
    if number <= 0:
        raise ValueError("Число должно быть пололжительным")
    count = 0
    while number > 0:
        count += number % 2
        number //= 2 
    return count 

test1 = 1
test2 = 12
test3 = 63
print(count_bit(test1) == 1)
print(count_bit(test2) == 2)
print(count_bit(test3) == 6)
```

# Задача 4
Написать функцию, которая принимает положительное число. Функция должна вернуть то,
сколько раз необходимо перемножать цифры числа или результат перемножения, чтобы
получилось число состоящее из одной цифры.
Например, для входного числа:
· 39 функция должна вернуть 3, так как 3*9=27 => 2*7=14 => 1*4=4
· 4 функция должна вернуть 0, так как число уже состоит из одной цифры
· 999 функция должна вернуть 4, так как 9*9*9=729 => 7*2*9=126 => 1*2*6=12 =>
1*2=2. Написать тесты для кода

```python
def magic(n):
    if n <= 0:
        raise ValueError("Число должно быть пололжительным")
    if n < 10: 
        return 0 
    count = 0
    current = n
    while current >= 10:
        process = 1
        for digit in str(current):
            process *= int(digit)
        current = process 
        count += 1
    return count 

test1 = 39
test2 = 999
test3 = 2
print(magic(test1) == 3)
print(magic(test2) == 4)
print(magic(test3) == 0)
```

# Задача 5
Написать функцию, которая принимает два целочисленных вектора одинаковой длины и
возвращает среднеквадратическое отклонение двух векторов. Написать тесты для кода

```python
def mse(pred, true):
    if len(pred) != len(true):
        raise ValueError("Векторы должны быть одинаковой длинны")
    lenght = len(pred)
    squared = 0
    for i in range(lenght):
        sqr = pred[i] - true[i]
        squared += sqr ** 2 
    return (squared / lenght) ** 0.5

print(mse([1, 2, 3], [1, 2, 3]) == 0)
print(mse([5], [3]) == 2)
print(mse([10, 20, 30], [5, 15, 25]) == 5)
```

# Задача 6
Написать функцию, принимающая целое положительное число. Функция должна вернуть
строку вида “(n1**p1)(n2**p2)…(nk**pk)” представляющая разложение числа на простые
множители (если pi == 1, то выводить только ni).
Например, для числа 86240 функция должна вернуть “(2**5)(5)(7**2)(11)”. Написать тесты для кода

```python
def prime_factors(num):
    if num <= 0:
        raise ValueError("Число должно быть пололжительным")
    if num == 1:
        return "(1)"
    divider = 2
    factors = []
    while divider * divider <= num:
        count = 0
        while num % divider == 0:
            count += 1
            num //= divider
        if count > 0:
            factors.append((divider, count))
        divider += 1
    
    if num > 1:
        factors.append((num, 1))
    
    result = []
    for factor, power in factors:
        if power == 1:
            result.append(f"({factor})")
        else:
            result.append(f"({factor}**{power})")
    
    return "".join(result)

print(prime_factors(123) == "(3)(41)")
print(prime_factors(86240) == "(2**5)(5)(7**2)(11)")
print(prime_factors(1) == "(1)")
```

# Задача 7
Написать функцию, принимающая целое число n, задающее количество кубиков. Функция
должна определить, можно ли из данного кол-ва кубиков построить пирамиду, то есть
можно ли представить число n как 1^2+2^2+3^2+…+k^2. Если можно, то функция должна
вернуть k, иначе строку “It is impossible”. Написать тесты для кода

```python
def pyramid(number):
    if number <= 0:
        return "It is impossible"
    k = 0
    result = 0
    while result < number:
        k += 1
        result += k * k
        if result == number:
            return k
    return "It is impossible"
print(pyramid(1) == 1)
print(pyramid(2) == "It is impossible")
print(pyramid(55) == 5)
```

# Задача 8
Функция принимает на вход положительное число и определяет является ли оно сбалансированным, т.е. сумма цифр до средних равна сумме цифр после. Средними в случае нечетного числа цифр считать одну цифру, в случае четного - две средних. Написать тесты для кода

```python
def balans(num):
    if num <= 0:
        raise ValueError("Число должно быть пололжительным")
    digit = str(num)
    lenght = len(digit)
    if lenght % 2 == 1:
        mid_start = lenght // 2
        mid_end = mid_start + 1
    else:
        mid_start = lenght // 2 - 1
        mid_end = mid_start + 2

    sum_before = sum(int(digit) for digit in digit[:mid_start])
    sum_after = sum(int(digit) for digit in digit[mid_end:])
    
    return sum_before == sum_after
print(balans(121) == True)
print(balans(66575) == True)
print(balans(11345) == False)
```
